```python
# =====================
# Dataset Overview
# =====================
total_population = 255.86
female_literacy = 65.19
youth = 74.2
middle_class = 45
lower_middle_class = 88
schooling = 45.90
intermediate = 2.56
university = 1.95
study_discontinued = 49
```

```python
# =====================
# Chart 1: Total Population Breakdown
# =====================
categories = [
    "Female Literacy Rate",
    "Youth (15–25)",
    "Middle Class",
    "Lower Middle Class",
    "Schooling",
    "Intermediate Education",
    "University Education",
    "Study Discontinued"
]
values = [
    female_literacy,
    youth,
    middle_class,
    lower_middle_class,
    schooling,
    intermediate,
    university,
    study_discontinued
]
```

```python
categories = [
    "Youth (15–25)", "Female Literacy", "Middle Class", "Lower Middle Class",
    "Education: Schooling", "Education: Intermediate", "Education: University",
    "Study Discontinued"
]

# Normalize values to avoid double-counting
# For example, only show education stats within youth or within middle class separately
```

```python
plt.figure(figsize=(5,5))
plt.pie(values, labels=[f"{c} ({v}M)" for c,v in zip(categories, values)], autopct='%1.1f%%', startangle=140)
plt.title("Total Population Breakdown (255.86M)")
plt.show()
```

```python
# =====================
# Chart 2: Youth vs Total Population
# =====================
remaining_population = total_population - youth
values = [youth, remaining_population]
labels = [f"Youth (15–25) ({youth}M)", f"Remaining ({remaining_population:.2f}M)"]

plt.figure(figsize=(5,5))
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=140, colors=["#66b3ff","#ff9999"])
plt.title("Youth vs Total Population")
plt.show()
```

```python
# =====================
# Chart 3: Youth Distribution in Classes
# =====================
middle_class_youth = (middle_class / (middle_class + lower_middle_class)) * youth
lower_middle_class_youth = (lower_middle_class / (middle_class + lower_middle_class)) * youth

values = [middle_class_youth, lower_middle_class_youth]
labels = [f"Middle Class Youth ({middle_class_youth:.2f}M)", f"Lower Middle Class Youth ({lower_middle_class_youth:.2f}M)"]

plt.figure(figsize=(6,6))
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=140, colors=["#99ff99","#ffcc99"])
plt.title("Youth Distribution in Middle and Lower Middle Class")
plt.show()
```

```python
# =====================
# Chart 4: Education Breakdown of Middle Class Youth
# =====================
# distribute education proportionally
total_education = schooling + intermediate + university + study_discontinued
mc_youth = middle_class_youth

mc_schooling = (schooling / total_education) * mc_youth
mc_intermediate = (intermediate / total_education) * mc_youth
mc_university = (university / total_education) * mc_youth
mc_discontinued = (study_discontinued / total_education) * mc_youth

values = [mc_schooling, mc_intermediate, mc_university, mc_discontinued]
labels = [
    f"Schooling ({mc_schooling:.2f}M)",
    f"Intermediate ({mc_intermediate:.2f}M)",
    f"University ({mc_university:.2f}M)",
    f"Study Discontinued ({mc_discontinued:.2f}M)"
]

plt.figure(figsize=(7,7))
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=140, colors=["#66b3ff","#99ff99","#ffcc99","#ff9999"])
plt.title("Education Breakdown of Youth in Middle Class")
plt.show()
```

```python
# =====================
# Chart 5: Education Breakdown of Lower Middle Class Youth
# =====================
lmc_youth = lower_middle_class_youth

lmc_schooling = (schooling / total_education) * lmc_youth
lmc_intermediate = (intermediate / total_education) * lmc_youth
lmc_university = (university / total_education) * lmc_youth
lmc_discontinued = (study_discontinued / total_education) * lmc_youth

values = [lmc_schooling, lmc_intermediate, lmc_university, lmc_discontinued]
labels = [
    f"Schooling ({lmc_schooling:.2f}M)",
    f"Intermediate ({lmc_intermediate:.2f}M)",
    f"University ({lmc_university:.2f}M)",
    f"Study Discontinued ({lmc_discontinued:.2f}M)"
]

plt.figure(figsize=(7,7))
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=140, colors=["#66b3ff","#99ff99","#ffcc99","#ff9999"])
plt.title("Education Breakdown of Youth in Lower Middle Class")
plt.show()
```

```python
import streamlit as st
import matplotlib.pyplot as plt

# =====================
# Dataset Overview
# =====================
total_population = 255.86
female_literacy = 65.19
youth = 74.2
middle_class = 45
lower_middle_class = 88
schooling = 45.90
intermediate = 2.56
university = 1.95
study_discontinued = 49

# =====================
# Helper Functions for Charts
# =====================

def chart1_total_population():
    categories = [
        "Female Literacy Rate",
        "Youth (15–25)",
        "Middle Class",
        "Lower Middle Class",
        "Schooling",
        "Intermediate Education",
        "University Education",
        "Study Discontinued"
    ]
    values = [
        female_literacy,
        youth,
        middle_class,
        lower_middle_class,
        schooling,
        intermediate,
        university,
        study_discontinued
    ]

    fig, ax = plt.subplots(figsize=(7,7))
    ax.pie(values, labels=[f"{c} ({v}M)" for c,v in zip(categories, values)], autopct='%1.1f%%', startangle=140)
    ax.set_title("Total Population Breakdown (255.86M)")
    st.pyplot(fig)

def chart2_youth_vs_population():
    remaining_population = total_population - youth
    values = [youth, remaining_population]
    labels = [f"Youth (15–25) ({youth}M)", f"Remaining ({remaining_population:.2f}M)"]

    fig, ax = plt.subplots(figsize=(6,6))
    ax.pie(values, labels=labels, autopct='%1.1f%%', startangle=140, colors=["#66b3ff","#ff9999"])
    ax.set_title("Youth vs Total Population")
    st.pyplot(fig)

def chart3_youth_distribution():
    middle_class_youth = (middle_class / (middle_class + lower_middle_class)) * youth
    lower_middle_class_youth = (lower_middle_class / (middle_class + lower_middle_class)) * youth

    values = [middle_class_youth, lower_middle_class_youth]
    labels = [f"Middle Class Youth ({middle_class_youth:.2f}M)", f"Lower Middle Class Youth ({lower_middle_class_youth:.2f}M)"]

    fig, ax = plt.subplots(figsize=(6,6))
    ax.pie(values, labels=labels, autopct='%1.1f%%', startangle=140, colors=["#99ff99","#ffcc99"])
    ax.set_title("Youth Distribution in Middle and Lower Middle Class")
    st.pyplot(fig)

def chart4_mc_education():
    total_education = schooling + intermediate + university + study_discontinued
    middle_class_youth = (middle_class / (middle_class + lower_middle_class)) * youth

    mc_schooling = (schooling / total_education) * middle_class_youth
    mc_intermediate = (intermediate / total_education) * middle_class_youth
    mc_university = (university / total_education) * middle_class_youth
    mc_discontinued = (study_discontinued / total_education) * middle_class_youth

    values = [mc_schooling, mc_intermediate, mc_university, mc_discontinued]
    labels = [
        f"Schooling ({mc_schooling:.2f}M)",
        f"Intermediate ({mc_intermediate:.2f}M)",
        f"University ({mc_university:.2f}M)",
        f"Study Discontinued ({mc_discontinued:.2f}M)"
    ]

    fig, ax = plt.subplots(figsize=(7,7))
    ax.pie(values, labels=labels, autopct='%1.1f%%', startangle=140,
           colors=["#66b3ff","#99ff99","#ffcc99","#ff9999"])
    ax.set_title("Education Breakdown of Youth in Middle Class")
    st.pyplot(fig)

def chart5_lmc_education():
    total_education = schooling + intermediate + university + study_discontinued
    lower_middle_class_youth = (lower_middle_class / (middle_class + lower_middle_class)) * youth

    lmc_schooling = (schooling / total_education) * lower_middle_class_youth
    lmc_intermediate = (intermediate / total_education) * lower_middle_class_youth
    lmc_university = (university / total_education) * lower_middle_class_youth
    lmc_discontinued = (study_discontinued / total_education) * lower_middle_class_youth

    values = [lmc_schooling, lmc_intermediate, lmc_university, lmc_discontinued]
    labels = [
        f"Schooling ({lmc_schooling:.2f}M)",
        f"Intermediate ({lmc_intermediate:.2f}M)",
        f"University ({lmc_university:.2f}M)",
        f"Study Discontinued ({lmc_discontinued:.2f}M)"
    ]

    fig, ax = plt.subplots(figsize=(7,7))
    ax.pie(values, labels=labels, autopct='%1.1f%%', startangle=140,
           colors=["#66b3ff","#99ff99","#ffcc99","#ff9999"])
    ax.set_title("Education Breakdown of Youth in Lower Middle Class")
    st.pyplot(fig)

# =====================
# Streamlit UI
# =====================

st.title("📊 Pakistan Education & Population Analysis Dashboard")

st.sidebar.header("Select a Chart")
chart_option = st.sidebar.radio(
    "Choose a chart to display:",
    ("Total Population Breakdown",
     "Youth vs Total Population",
     "Youth Distribution in Classes",
     "Education Breakdown of Middle Class Youth",
     "Education Breakdown of Lower Middle Class Youth")
)

if chart_option == "Total Population Breakdown":
    chart1_total_population()
elif chart_option == "Youth vs Total Population":
    chart2_youth_vs_population()
elif chart_option == "Youth Distribution in Classes":
    chart3_youth_distribution()
elif chart_option == "Education Breakdown of Middle Class Youth":
    chart4_mc_education()
elif chart_option == "Education Breakdown of Lower Middle Class Youth":
    chart5_lmc_education()
```
